```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import plotly.graph_objects as go

# --- 1️⃣ Charger les données ---
scaled_data = pd.read_csv(
    'C:/Users/Mega-Pc/Freelance/data/ecommerce/data_cleaned+scaled.csv',
    encoding='ISO-8859-1'
)

real_data = pd.read_csv(
    'C:/Users/Mega-Pc/Freelance/data/ecommerce/data_before_scaling.csv',
    encoding='ISO-8859-1'
)

# --- 2️⃣ Agréger les deux par CustomerID ---
def aggregate_by_customer(df):
    return df.groupby('CustomerID').agg({
        'TotalSpent': 'sum',
        'NumTransactions': 'sum',
        'AvgTransactionValue': 'mean',
        'MaxTransactionValue': 'max',
        'TotalPrice': 'sum',
        'AvgPricePerItem': 'mean',
        'AvgUnitPriceSold': 'mean',
        'TotalQuantity': 'sum',
        'Quantity': 'sum',
        'AvgItemsPerTransaction': 'mean',
        'NumUniqueItemsBought': 'sum',
        'CustomerProductAffinity': 'mean',
        'RecencyDays': 'mean',
        'FrequencyPerMonth': 'mean',
        'MonetaryPerMonth': 'mean',
        'AvgDaysBetweenInvoices': 'mean',
        'TenureDays': 'mean',
        'TenureMonths': 'mean',
        'Recency_x_Frequency': 'mean',
        'Frequency_x_Monetary': 'mean',
        'ReturnRate': 'mean',
        'SentimentScore': 'mean',
        'AvgSentimentScore': 'mean',
        'IsDeluxe': 'mean',
        'InvoiceMonth': lambda x: x.mode()[0],
        'InvoiceQuarter': lambda x: x.mode()[0],
        'InvoiceDayOfWeek': lambda x: x.mode()[0],
        'InvoiceHour': lambda x: x.mode()[0],
        'IsWeekend': 'mean',
        'IsHolidaySeason': 'mean',
        'CountryAvgSpend': 'mean',
        'CountryAvgSpendDeviation': 'mean',
        'FirstPurchase_Year': 'min',
        'FirstPurchase_Month': 'min',
        'FirstPurchase_Day': 'min',
        'FirstPurchase_Hour': 'min',
        'LastPurchase_Year': 'max',
        'LastPurchase_Month': 'max',
        'LastPurchase_Day': 'max',
        'LastPurchase_Hour': 'max',
    }).reset_index()

scaled_data = aggregate_by_customer(scaled_data)
real_data = aggregate_by_customer(real_data)

# --- 3️⃣ Définir les features ---
features = [
    'TotalSpent', 'NumTransactions', 'AvgTransactionValue', 'MaxTransactionValue',
    'TotalQuantity', 'AvgItemsPerTransaction', 'NumUniqueItemsBought', 'CustomerProductAffinity',
    'RecencyDays', 'FrequencyPerMonth', 'MonetaryPerMonth', 'AvgDaysBetweenInvoices',
    'TenureDays', 'TenureMonths', 'Recency_x_Frequency', 'Frequency_x_Monetary',
    'ReturnRate', 'SentimentScore', 'AvgSentimentScore', 'IsDeluxe',
    'InvoiceMonth', 'InvoiceQuarter', 'InvoiceDayOfWeek', 'InvoiceHour',
    'IsWeekend', 'IsHolidaySeason', 'CountryAvgSpendDeviation'
]

X_scaled = scaled_data[features]

# --- 4️⃣ Chercher K optimal ---
scores = []
K_range = range(2, 10)

for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X_scaled)
    score = silhouette_score(X_scaled, kmeans.labels_)
    scores.append(score)

optimal_k = K_range[scores.index(max(scores))]
optimal_score = max(scores)

print(f"K optimal = {optimal_k} ")

# --- 5️⃣ Appliquer KMeans final ---
kmeans = KMeans(n_clusters=optimal_k, random_state=42)
real_data['Cluster'] = kmeans.fit_predict(X_scaled)

# --- 6️⃣ Profil des clusters ---
cluster_profile = real_data.groupby('Cluster')[features].mean().round(2)
print(cluster_profile)
```

```python
features = [
    # Dépenses et valeur transactionnelle
    'TotalSpent',
    'NumTransactions',
    'AvgTransactionValue',
    'MaxTransactionValue',
    'TotalQuantity',
    'AvgItemsPerTransaction',
    'NumUniqueItemsBought',
    'CustomerProductAffinity',
    
    # RFM enrichi
    'RecencyDays',
    'FrequencyPerMonth',
    'MonetaryPerMonth',
    'AvgDaysBetweenInvoices',
    'TenureDays',
    'TenureMonths',
    'Recency_x_Frequency',
    'Frequency_x_Monetary',
    
    # Retour et satisfaction
    'ReturnRate',
    'SentimentScore',
    'AvgSentimentScore',
    'IsDeluxe',
    
    # Timing / saisonnalité (si tu veux inclure les patterns temporels)
    'InvoiceMonth',
    'InvoiceQuarter',
    'InvoiceDayOfWeek',
    'InvoiceHour',
    'IsWeekend',
    'IsHolidaySeason',
    
    # Géographie / marché
    'CountryAvgSpendDeviation'
]


X = customer_df[features]
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import plotly.graph_objects as go


scores = []
K_range = range(2, 10)

for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X)  
    score = silhouette_score(X, kmeans.labels_)
    scores.append(score)


optimal_k = K_range[np.argmax(scores)]
optimal_score = max(scores)

fig = go.Figure()


fig.add_trace(go.Scatter(
    x=list(K_range),
    y=scores,
    mode='lines+markers',
    name='Silhouette Score',
    line=dict(color='royalblue', width=3),
    marker=dict(size=8)
))


fig.add_trace(go.Scatter(
    x=[optimal_k],
    y=[optimal_score],
    mode='markers+text',
    name=f'K optimal = {optimal_k}',
    text=[f'K={optimal_k}'],
    textposition="top center",
    marker=dict(color='red', size=12, symbol='x')
))


fig.update_layout(
    title="📈 Silhouette Score selon le nombre de clusters (K)",
    xaxis_title="Nombre de clusters (K)",
    yaxis_title="Silhouette Score",
    template="plotly_white",
    font=dict(size=14),
    showlegend=True
)

fig.show()

print(f"K optimal = {optimal_k} avec un score de silhouette de {optimal_score:.3f}")
```

# 🎯 Analyse des Segments Clients et Recommandations Marketing

Après avoir segmenté la base clients, deux groupes principaux se dégagent avec des comportements bien distincts :

---

## 🔥 Cluster 0 : Les Clients Premium / Deluxe

### Profil
- 💰 Dépenses totales très élevées  
- 📈 Achats fréquents et réguliers  
- 🛒 Panier moyen plus important que la moyenne  
- ✨ Forte présence de clients “Deluxe”  
- 😊 Sentiment global positif et fort attachement à la marque  

### Ce qu’il faut en retenir
Ces clients sont la **source principale de chiffre d’affaires**.  
Ils achètent souvent, dépensent plus et montrent une vraie fidélité.  
Ce sont des clients à **forte valeur**, qu’il faut continuer à choyer et à engager.

### Recommandations
- Mettre en place un **programme de fidélité premium** (accès prioritaire, récompenses exclusives, offres VIP).  
- Proposer des **promotions personnalisées** sur des produits haut de gamme.  
- Maintenir une **relation directe et valorisante** via email ou SMS.  
- Travailler sur des **stratégies d’upsell et de cross-sell** pour renforcer la rentabilité.  

---

## 🌱 Cluster 1 : Les Clients Occasionnels / Standard

### Profil
- 💸 Dépenses globales plus faibles  
- 📉 Fréquence d’achat irrégulière  
- 🛒 Panier moyen plus petit  
- 😐 Engagement et fidélité plus faibles  

### Ce qu’il faut en retenir
Ces clients achètent moins souvent et ont un **potentiel de croissance**.  
Ils ne sont pas perdus, mais nécessitent une **approche de réactivation**.  
Souvent sensibles aux **offres attractives** et aux **incitations à revenir**.

### Recommandations
- Relancer ces clients avec des **offres limitées dans le temps** ou des **remises personnalisées**.  
- Encourager la fidélité avec un **système de points ou récompenses simples**.  
- Cibler les campagnes selon leurs **intérêts ou historique d’achat**.  
- Créer un lien plus fort avec la marque à travers des messages clairs et humains.  

---

## ✅ En résumé
- **Cluster 0 →** Clients premium : à fidéliser et à valoriser avec des offres exclusives.  
- **Cluster 1 →** Clients occasionnels : à réactiver et à motiver pour les transformer en fidèles.  

> Cette segmentation permet de concentrer les efforts sur les leviers les plus rentables : **renforcer la relation avec les meilleurs clients** tout en **redynamisant les segments moins actifs**.
